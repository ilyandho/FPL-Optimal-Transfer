```python
# FantasyGo Automation Scraper (Jupyter Notebook Version)
# Purpose: Fetch EPL Gameweeks -> Get Leaderboards -> Fetch Team Selections

import requests
import json
import time
import pandas as pd
from typing import List, Dict

# --- 1. CONFIGURATION & AUTHENTICATION ---
# Run this cell and paste your token when prompted
raw_token = input("Please paste your Bearer Token from the browser: ")
auth_token = raw_token.replace("Bearer ", "").strip()

base_url = "https://api.fantasygo.io/public"
headers = {
    "Content-Type": "application/json",
    "Authorization": f"Bearer {auth_token}",
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/143.0.0.0 Safari/537.36",
    "Origin": "https://app.fantasygo.io",
    "Referer": "https://app.fantasygo.io/"
}

# --- 2. STEP 1: FETCH EPL GAMEWEEKS ---
print("--- STEP 1: Fetching EPL Gameweeks ---")

ended_contests_query = """
query EndedContests($args: EndedContestsInput) {
  ended_contests: endedContests(args: $args) {
    contests {
      id
      name
      isSingleMatchContest
    }
  }
}
"""

variables = {
    "args": {
        "eventTypes": ["englishPremierLeague"],
        "gameTypes": ["fantasy"],
        "enteredOnly": False,
        "limit": 50,
        "offset": 0
    }
}

response = requests.post(base_url, headers=headers, json={
    "operationName": "EndedContests",
    "query": ended_contests_query,
    "variables": variables
})

if response.status_code != 200 or "errors" in response.json():
    print(f"Error fetching contests: {response.text}")
else:
    all_contests = response.json()['data']['ended_contests']['contests']
    # Filter for Gameweeks only
    gameweeks = [c for c in all_contests if not c['isSingleMatchContest']]
    print(f"Found {len(gameweeks)} Gameweek contests.")

# --- 3. LOOPS: PROCESS EACH GAMEWEEK ---
all_extracted_data = []

for gw in gameweeks:
    print(f"\nProcessing: {gw['name']} (ID: {gw['id']})")

    # STEP 2: FETCH LEADERBOARD
    leaderboard_query = """
    query LeaderboardForContest($args: LeaderboardForContestInput!) {
      leaderboard: leaderboardForContest(args: $args) {
        items: contestLeaderboardItems {
          userContestId
          username
          rank
          totalPoints
        }
      }
    }
    """

    lb_vars = {
        "args": {
            "contestId": int(gw['id']),
            "limit": 10, # Top 10 users
            "offset": 0,
            "entryFee": 10
        }
    }

    lb_res = requests.post(base_url, headers=headers, json={
        "operationName": "LeaderboardForContest",
        "query": leaderboard_query,
        "variables": lb_vars
    }).json()

    entries = lb_res['data']['leaderboard']['items']
    gw_results = []

    # STEP 3: FETCH INDIVIDUAL TEAM SELECTIONS
    for entry in entries:
        print(f"  Fetching selection for Rank {entry['rank']}: {entry['username']}...")

        team_query = """
        query UserContest($args: GetUserContestInput!) {
          user_contest: userContest(args: $args) {
            team: userContestTeam {
              formation
              players: userContestPlayers {
                eventTeamPlayerId
                isCaptain
                totalPoints
              }
            }
          }
        }
        """

        team_vars = {"args": {"id": int(entry['userContestId'])}}

        try:
            team_res = requests.post(base_url, headers=headers, json={
                "operationName": "UserContest",
                "query": team_query,
                "variables": team_vars
            }).json()

            user_team_data = team_res['data']['user_contest']['team']

            full_entry = {
                "Gameweek": gw['name'],
                "Rank": entry['rank'],
                "Username": entry['username'],
                "Points": entry['totalPoints'],
                "Formation": user_team_data['formation'],
                "Players": user_team_data['players']
            }
            gw_results.append(full_entry)
            all_extracted_data.append(full_entry)

        except Exception as e:
            print(f"    Error fetching team for {entry['username']}: {e}")

        # Small delay to avoid Cloudflare rate limiting
        time.sleep(0.3)

    # Save individual Gameweek to JSON
    filename = f"Result_{gw['name'].replace(' ', '_')}.json"
    with open(filename, 'w') as f:
        json.dump(gw_results, f, indent=4)
    print(f"Saved results to {filename}")

# --- 4. FINAL SUMMARY IN PANDAS ---
print("\n--- Automation Complete! ---")
df = pd.DataFrame(all_extracted_data)
# Displays the first few rows of the scraped data in the notebook
df.head()




```
