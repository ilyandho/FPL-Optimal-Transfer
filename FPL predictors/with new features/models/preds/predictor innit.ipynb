# Utils


```python
def calc_xP_pred(row):
    # Minutes played bonus
    if row['minutes_3'] >= 60:
        minutes_bonus = 2
    elif row['minutes_3'] > 0:
        minutes_bonus = 1
    else:
        minutes_bonus = 0  # if player didn't play at all
    if row['position'] == '1':

        ## GK
        ## ---> >60  = 2
        ## ---> cs*4
        ## ---> ((sv)/3)*1
        ## ---> (gc/2)*-1
        ## ---> yc*-1
        return (
            minutes_bonus +
            4 * row['cs_preds'] +
            1 * round(row['saves_preds']/3, 3) -
            1 * round(row['xgc_preds']/2, 3) -
            1 * row['y_preds']
        )

    elif row['position'] == '2':
        ## DEF
        ## ---> >60  = 2
        ## ---> xg*6
        ## ---> xa*3
        ## ---> cs*4
        ## ---> gc
        ## ---> yc*-1


        return (
            minutes_bonus +
            6 * row['xg_preds'] +
            3 * row['xa_preds'] +
            4 * row['cs_preds'] -
            1 * round(row['xgc_preds']/2, 3) -
            1 * row['y_preds']
        )
    elif row['position'] == '3':
        ## MID
        ## ---> >60  = +2
        ## ---> xg*5
        ## ---> xa*3
        ## ---> cs*1
        ## ---> yc*-1

        return (
            minutes_bonus +
            5 * row['xg_preds'] +
            3 * row['xa_preds'] +
            1 * row['cs_preds'] -
            1 * row['y_preds']
        )
    elif row['position'] == '4':
        ## FWD
        ## ---> >60  = 2
        ## ---> xg*4
        ## ---> xa*3
        ## ---> yc*-1
        # print(row['fpl_name'], row['position'], minutes_bonus +
        #     4 * row['xg_preds'] +
        #     3 * row['xa_preds'] -
        #     1 * row['y_preds'])
        return (
            minutes_bonus +
            4 * row['xg_preds'] +
            3 * row['xa_preds'] -
            1 * row['y_preds']
        )
    else:
        return 0  # fallback for unknown positions
```
