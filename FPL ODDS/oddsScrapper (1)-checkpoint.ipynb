```python
# function to get unique values
def unique(list1):
 
    # initialize a null list
    unique_list = []
     
    # traverse for all elements
    for x in list1:
        # check if exists in unique_list or not
        if x not in unique_list:
            unique_list.append(x)
    # print list
    return(unique_list)
```

```python
# Get all the match day urls
# match_links = sel.xpath('//*[@id="oddsshark-upcoming-events"]/div/div/div[2]/ul/li/div/a/')
# match_links = sel.xpath('//*[@id="oddsshark-upcoming-events"]/div/div/div[2]/ul/li/div/a[contains(@href, "/soccer/epl/")]/@href').getall()
match_links = sel.xpath('//*[@id="oddsshark-ticker"]/div/div/div[2]/div/div/div/div/a[contains(@href, "/soccer/epl/")]/@href').getall()

# retain unique urls from each pair
# unique_match_links = unique([i.get() for i in match_links])

# add the prefix
full_links = ['https://www.oddsshark.com' + i for i in match_links]
full_links
```

```python
rows_list = []
for i in full_links:
    # to get urls for respective matches
    driver.get(i)
    
    sleep(2)
    
    # grab the webpage
    sel2 = Selector(text = driver.page_source)
    
    # Extract the names for the teams playing
#     teams = sel2.xpath('//*[starts-with(@class, "gc-team-info__text--primary")]/text()').extract()
    home_team = sel2.xpath('//*[starts-with(@class, "gc-both-home-name")]/text()').extract()[0]
    away_team = sel2.xpath('//*[starts-with(@class, "gc-both-away-name")]/text()').extract()[0]
    
    print(home_team+' v '+ away_team)
    
    # Extract the historic performance data
#     form = sel2.xpath('//*[starts-with(@class, "gc-team-info__text--secondary")]/text()').extract()
    away_perfomance = sel2.xpath('//*[starts-with(@class, "gc-both-away-subtitle-one")]/text()').extract()[0]
    away_form =  sel2.xpath('//*[starts-with(@class, "gc-both-away-subtitle-two")]/text()').extract()[0]
    
    home_perfomance = sel2.xpath('//*[starts-with(@class, "gc-both-home-subtitle-one")]/text()').extract()[0]
    home_form = sel2.xpath('//*[starts-with(@class, "gc-both-home-subtitle-two")]/text()').extract()[0]
    
    # Extract the predicted scores
    home_score = sel2.xpath('//*[starts-with(@class, "gc-score__num gc-score__num--left")]/div/text()').extract_first()
    away_score = sel2.xpath('//*[starts-with(@class, "gc-score__num gc-score__num--right")]/div/text()').extract_first()

    # store all the collected match data into a dictionary
    dict1 = {'home_team': home_team, 'away_team':away_team,'home_score':home_score,'away_score':away_score,'home_form':home_form, 'away_form':away_form, 'home_perfomance':home_perfomance, 'away_perfomance':away_perfomance}
    rows_list.append(dict1)
    
    sleep(2)
```

```python
df = pd.DataFrame(rows_list)
df
```

```python
df = pd.DataFrame(rows_list)
df
```

```python
df.to_csv(file_name, index = False)# encoding = 'utf-8'
```

# William Hill

```python
driver.get('https://sports.williamhill.com/betting/en-gb/football/competitions/OB_TY295/English-Premier-League/matches/OB_MGMB/Match-Betting')
```

```python
sel = Selector(text = driver.page_source)

match_links = sel.xpath('//*[starts-with(@class, "sp-o-market__more")]/a/@href')

# Extract the urls from the Selector object
unique_match_links = ['https://sports.williamhill.com'+i.get() for i in match_links]

unique_match_links
```

```python
unique_match_links
```

```python
appended_data = []
```

```python
//*[@id="page"]/div[1]/div/main/div/article/div/h1
```

```python

for i in range(10,12):
    driver.get(unique_match_links[i])

    sleep(3)
    # grab the webpage
    sel2 = Selector(text = driver.page_source)

    contest = sel2.xpath('//*[starts-with(@class, "css-qbolbz")]/text()').extract()
    teams = contest[0].split(" v ")
    
    player_name = sel2.xpath('//*[starts-with(@class, "cpt-goalscorer__title")]/h3/text()').extract()
    anytime = sel2.xpath('//li[starts-with(@class, "odds-market-2")]/button/@data-odds')
    two_or_more = sel2.xpath('//li[starts-with(@class, "odds-market-4")]/button/@data-odds')
    hatrick = sel2.xpath('//li[starts-with(@class, "odds-market-5")]/button/@data-odds')

    anytime_odds = [i.get() for i in anytime]
    two_or_more = [i.get() for i in two_or_more]
    hatrick = [i.get() for i in hatrick]

    player_name= player_name[:-1]
    
    print(contest[0])
    print(len(player_name),len(anytime_odds), len(hatrick), len(two_or_more))

    wh = pd.DataFrame({'home_team': teams[0], 'away_team': teams[1],'player_name':player_name, 'anytime_odds':anytime_odds, 'two_or_more':two_or_more, 'hatrick':hatrick })

    appended_data.append(wh)
    
```
