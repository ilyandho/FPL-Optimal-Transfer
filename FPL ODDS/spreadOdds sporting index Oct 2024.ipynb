```python
# function to get unique values
def unique(list1):

    # initialize a null list
    unique_list = []

    # traverse for all elements
    for x in list1:
        # check if exists in unique_list or not
        if x not in unique_list:
            unique_list.append(x)
    # print list
    return(unique_list)
```

```python
# Function for extracting
def get_full_xpath(element):
    components = []
    child = element
    while child.tag_name != "html":
        parent = child.find_element(By.XPATH, "..")
        siblings = parent.find_elements(By.XPATH, "*")
        same_tag_siblings = [sib for sib in siblings if sib.tag_name == child.tag_name]

        # Attempt to use a more stable identifier if available
        data_id = child.get_attribute("data-id")
        if data_id:
            position = f"[@data-id='{data_id}']"
        elif len(same_tag_siblings) > 1:
            index = same_tag_siblings.index(child) + 1
            position = f"[{index}]"
        else:
            position = ''

        components.append(f"{child.tag_name}{position}")
        child = parent

    components.append("html")
    components.reverse()
    return '/' + '/'.join(components)
```

```python
button_elements = driver.find_elements(By.XPATH, button_xpath)
max_retries = 3
assist_index = 3

for i in range(3,10):
    print('On: '+contest_list[i])
    retries = 0
    while retries < max_retries:
        try:
            # Re-find the elements to handle staleness
            button_elements = WebDriverWait(driver, 10).until(
                EC.presence_of_all_elements_located((By.XPATH, button_xpath))
            )
            # Click the element
            button_elements[i].click()
            # Optional: Wait for a moment
            time.sleep(3)

            break  # Exit the loop if click is successful

        except StaleElementReferenceException:
            print("Element reference is stale. Retrying...")
            retries += 1
    if retries == max_retries:
        print(f"Failed to click button at index {i} after {max_retries} retries.")

    time.sleep(5)


    #     ############################################################
    #     #scorer click
    #     ############################################################
    # After clicking the buttons, find buttons with the specified text
    target_text = "Player Goal Minutes"
    # XPath selector for buttons with the target text
    target_buttons_xpath = f'//*[contains(text(), "{target_text}")]'
    # Find all buttons with the target text
    target_buttons1 = driver.find_elements(By.XPATH, target_buttons_xpath)

    for button in target_buttons1:
        try:
            # Wait up to 10 seconds for the button to be clickable
            WebDriverWait(driver, 10).until(EC.element_to_be_clickable((By.XPATH, target_buttons_xpath)))
            # Scroll into view
            driver.execute_script("arguments[0].scrollIntoView({block: 'center'});", button)

            # Try normal click first
            try:
                button.click()
            except Exception:
                # Fallback: JS click
                driver.execute_script("arguments[0].click();", button)
            # Optional: Wait for a moment
            time.sleep(3)
        except TimeoutException:
            print(f"Timeout waiting for the button to be clickable.")


    #     ############################################################
    #     #PLAYERS click
    #     ############################################################
    # After clicking the buttons, find buttons with the specified text
    target_text = "Players"
    # XPath selector for buttons with the target text
    target_buttons_xpath = f'//*[contains(text(), "{target_text}")]'
    # Find all buttons with the target text
    target_buttons = driver.find_elements(By.XPATH, target_buttons_xpath)
    # Print or interact with the found buttons
    try:
        # Wait up to 10 seconds for the button to be clickable
        WebDriverWait(driver, 10).until(EC.element_to_be_clickable((By.XPATH, target_buttons_xpath)))
        # Scroll the element into view
        ActionChains(driver).move_to_element(target_buttons[0]).perform()
        # Click the button using JavaScript
        driver.execute_script("arguments[0].click();", target_buttons[0])
        # Optional: Wait for a moment
        time.sleep(3)
    except TimeoutException:
        print(f"Timeout waiting for the button to be clickable.")


    ############################################################
    #Scorer Markets
    ############################################################
    time.sleep(4)
    matches = []

    for button in target_buttons1:
        full_xpath = get_full_xpath(button)
        # Regular expression pattern to find "container[<number>]"
        pattern = r"container\[(\d+)\]"
        # Search for all occurrences and extract numbers
        matches.append(re.findall(pattern, full_xpath))

    home_index = matches[0]
    away_index = matches[1]

    # HOME DATA
    init_path = '//sx-pricing-page//sx-p-panel-container['+ home_index[0]
    target_xpath_pattern = init_path + ']//sx-p-panel-spreads-column[1]/div/div[*]/div[2]/div/div/sx-p-panel-link-name/div/span'
    # /sc-panel/div/div[2]/content/sx-p-panel-spreads/div/div/sx-p-panel-spreads-column[1]/div/div[2]/div[2]/div/div/sx-p-panel-link-name/div/span
    # sx-pricing-page/div[1]/sc-panel/div/div[2]/content/sx-p-panel-container[1]/div/sc-panel/div/div[2]/content/sx-p-panel-spreads/div/div/sx-p-panel-spreads-column[1]/div/div[2]/div[2]/div/div/sx-p-panel-link-name/div/span
    # /html/body/app/div/main/sx-content/div/div[3]/div[3]/
    retries = 0
    while retries < max_retries:
        try:
            # Find all elements that match the specified XPath pattern
            target_elements = driver.find_elements(By.XPATH, target_xpath_pattern)
            # Extract and print information from all matching elements
            PGM_names = [target_element.text for target_element in target_elements]
            # Click the element
            1/len(PGM_names)

            break  # Exit the loop if click is successful

        except ZeroDivisionError:
            print("Division by zero. Retrying...")
            retries += 1
            # Optional: Wait for a moment
            time.sleep(3)


    # After clicking the initial buttons, find elements that match the specified XPath pattern
    target_xpath_pattern = init_path + ']//sx-p-panel-spreads-column[1]/div/div[*]/div[3]/sx-spr-price-button/div/div'
    # Find all elements that match the specified XPath pattern
    target_elements = driver.find_elements(By.XPATH, target_xpath_pattern)
    # Extract and print information from all matching elements
    PGM_odds = [target_element.text for target_element in target_elements]

    # After clicking the initial buttons, find elements that match the specified XPath pattern
    target_xpath_pattern = init_path + ']//sx-p-panel-spreads-column[2]/div/div[*]/div[2]/div/div/sx-p-panel-link-name/div/span'
    # Find all elements that match the specified XPath pattern
    target_elements = driver.find_elements(By.XPATH, target_xpath_pattern)
    # Extract and print information from all matching elements
    SMM_names = [target_element.text for target_element in target_elements]

    # After clicking the initial buttons, find elements that match the specified XPath pattern
    target_xpath_pattern = init_path + ']//sx-p-panel-spreads-column[2]/div/div[*]/div[3]/sx-spr-price-button/div/div'
    # Find all elements that match the specified XPath pattern
    target_elements = driver.find_elements(By.XPATH, target_xpath_pattern)
    # Extract and print information from all matching elements
    SMM_odds = [target_element.text for target_element in target_elements]

    home_df = pd.DataFrame({"PGM_names": PGM_names, "PGM_odds": PGM_odds,"SMM_names": SMM_names, "SMM_odds": SMM_odds})

    print('home PGM: '+ str(len(PGM_names)),'home SMM: '+ str(len(SMM_names)))

    # AWAY DATA
    # After clicking the initial buttons, find elements that match the specified XPath pattern
    init_path = '//sx-pricing-page//sx-p-panel-container['+ away_index[0]

    target_xpath_pattern = init_path + ']//sx-p-panel-spreads-column[1]/div/div[*]/div[2]/div/div/sx-p-panel-link-name/div/span'

    retries = 0
    while retries < max_retries:
        try:
            # Find all elements that match the specified XPath pattern
            target_elements = driver.find_elements(By.XPATH, target_xpath_pattern)
            # Extract and print information from all matching elements
            PGM_names = [target_element.text for target_element in target_elements]

            # Click the element
            1/len(PGM_names)

            break  # Exit the loop if click is successful

        except ZeroDivisionError:
            print("Division by zero. Retrying...")
            retries += 1
            # Optional: Wait for a moment
            time.sleep(3)

    # After clicking the initial buttons, find elements that match the specified XPath pattern
    target_xpath_pattern = init_path + ']//sx-p-panel-spreads-column[1]/div/div[*]/div[3]/sx-spr-price-button/div/div'
    # Find all elements that match the specified XPath pattern
    target_elements = driver.find_elements(By.XPATH, target_xpath_pattern)
    # Extract and print information from all matching elements
    PGM_odds = [target_element.text for target_element in target_elements]

    # After clicking the initial buttons, find elements that match the specified XPath pattern
    target_xpath_pattern = init_path + ']//sx-p-panel-spreads-column[2]/div/div[*]/div[2]/div/div/sx-p-panel-link-name/div/span'
    # Find all elements that match the specified XPath pattern
    target_elements = driver.find_elements(By.XPATH, target_xpath_pattern)
    # Extract and print information from all matching elements
    SMM_names = [target_element.text for target_element in target_elements]

    # After clicking the initial buttons, find elements that match the specified XPath pattern
    target_xpath_pattern = init_path + ']//sx-p-panel-spreads-column[2]/div/div[*]/div[3]/sx-spr-price-button/div/div'
    # Find all elements that match the specified XPath pattern
    target_elements = driver.find_elements(By.XPATH, target_xpath_pattern)
    # Extract and print information from all matching elements
    SMM_odds = [target_element.text for target_element in target_elements]

    print('away PGM: '+ str(len(PGM_names)),'away SMM: '+ str(len(SMM_names)))

    away_df = pd.DataFrame({"PGM_names": PGM_names, "PGM_odds": PGM_odds,"SMM_names": SMM_names, "SMM_odds": SMM_odds})

    contest_df = pd.concat([home_df, away_df], ignore_index=True)
    contest_df["contest"] = contest_list[i]

    contest_df_list.append(contest_df)

    print('Done with Scorers')


    #     ############################################################
    #     #Player Performance Points
    #     ############################################################
    # After clicking the buttons, find buttons with the specified text
    target_text = "Player Performance Points"
    # XPath selector for buttons with the target text
    target_buttons_xpath = f'//*[contains(text(), "{target_text}")]'
    # Find all buttons with the target text
    target_buttons4 = driver.find_elements(By.XPATH, target_buttons_xpath)

    matches = []
    # HOME DATA
    for button in target_buttons4:
        full_xpath = get_full_xpath(button)

        # Regular expression pattern to find "container[<number>]"
        pattern = r"container\[(\d+)\]"

        # Search for all occurrences and extract numbers
        matches = re.findall(pattern, full_xpath)

    if len(matches)!=0:
        # Print extracted numbers
        for match in matches:
            perfomance_index = match

        init_path = '//sx-pricing-page//sx-p-panel-container['+ str(perfomance_index)
        #                sx-pricing-page/div[1]/sc-panel/div/div[2]/content/sx-p-panel-container[4]/div/sc-panel/div/div/div[2]/header/div/div[1]/span
        target_xpath_pattern = init_path + ']//sx-p-panel-spreads-column[*]/div/div[*]/div[2]/div/div/sx-p-panel-link-name/div/span'
        retries = 0
        while retries < max_retries:
            try:
                # Find all elements that match the specified XPath pattern
                target_elements = driver.find_elements(By.XPATH, target_xpath_pattern)
                # Extract and print information from all matching elements
                perfomance_names = [target_element.text for target_element in target_elements]

                # Click the element
                1/len(perfomance_names)

                break  # Exit the loop if click is successful

            except ZeroDivisionError:
                print("Zero names Found. Retrying...")
                retries += 1
                # Optional: Wait for a moment
                time.sleep(3)
        perfomance_names

        # After clicking the initial buttons, find elements that match the specified XPath pattern
        target_xpath_pattern = init_path + ']//sx-p-panel-spreads-column[*]/div/div[*]/div[3]/sx-spr-price-button/div/div'
        # Find all elements that match the specified XPath pattern
        target_elements = driver.find_elements(By.XPATH, target_xpath_pattern)
        # Extract and print information from all matching elements
        perfomance_odds = [target_element.text for target_element in target_elements]

        perfomance_df = pd.DataFrame({"perfomance_names": perfomance_names, "perfomance_odds": perfomance_odds})

        perfomance_df = pd.concat([perfomance_df], ignore_index=True)
        perfomance_df['contest'] = contest_list[i]
        perfomance_df_list.append(perfomance_df)

        print('Done with Perfomance Points')
    else:
        print('Perfomance Points not Found')
#     ############################################################
#     #Cards click
#     ############################################################
    # After clicking the buttons, find buttons with the specified text
    target_text = "Cards"
    # XPath selector for buttons with the target text
    target_buttons_xpath = f'//*[contains(text(), "{target_text}")]'
    # Find all buttons with the target text
    target_buttons6 = driver.find_elements(By.XPATH, target_buttons_xpath)
    # Print or interact with the found buttons

    # After clicking the buttons, find buttons with the specified text
    target_text = "Card Markets"
    # XPath selector for buttons with the target text
    target_buttons_xpath = f'//*[contains(text(), "{target_text}")]'
    # Find all buttons with the target text
    target_buttons6 = driver.find_elements(By.XPATH, target_buttons_xpath)

    if len(target_buttons6)!=0:
        time.sleep(4)
        matches = []

        for button in target_buttons6:
            full_xpath = get_full_xpath(button)
            # Regular expression pattern to find "container[<number>]"
            pattern = r"container\[(\d+)\]"
            # Search for all occurrences and extract numbers
            matches.append(re.findall(pattern, full_xpath))

        home_cards_index = matches[0]
        away_cards_index = matches[1]


        # After clicking the Cards button
        init_path = '//sx-pricing-page//sx-p-panel-container['+ str(home_cards_index[0])
        target_xpath_pattern = init_path + ']//sx-p-panel-spreads-column[*]/div/div[*]/div[2]/div/div/sx-p-panel-link-name/div/span'
        retries = 0

        while retries < max_retries:
            try:
                # Find all elements that match the specified XPath pattern
                target_elements = driver.find_elements(By.XPATH, target_xpath_pattern)
                # Extract and print information from all matching elements
                home_card_names = [target_element.text for target_element in target_elements]

                # Click the element
                1/len(home_card_names)

                break  # Exit the loop if click is successful

            except ZeroDivisionError:
                print("Zero names Found. Retrying...")
                retries += 1
                # Optional: Wait for a moment
                time.sleep(3)
        # After clicking the initial buttons, find elements that match the specified XPath pattern
        target_xpath_pattern = init_path + ']//sx-p-panel-spreads-column[*]/div/div[*]/div[3]/sx-spr-price-button/div/div'
        target_elements = driver.find_elements(By.XPATH, target_xpath_pattern)
        # Extract and print information from all matching elements
        home_card_odds = [target_element.text for target_element in target_elements]

        init_path = '//sx-pricing-page//sx-p-panel-container['+ (away_cards_index[0])
        # After clicking the initial buttons, find elements that match the specified XPath pattern
        target_xpath_pattern = init_path + ']//sx-p-panel-spreads-column[*]/div/div[*]/div[2]/div/div/sx-p-panel-link-name/div/span'

        target_elements = driver.find_elements(By.XPATH, target_xpath_pattern)
        # Extract and print information from all matching elements
        away_card_names = [target_element.text for target_element in target_elements]

        # After clicking the initial buttons, find elements that match the specified XPath pattern
        target_xpath_pattern = init_path + ']//sx-p-panel-spreads-column[*]/div/div[*]/div[3]/sx-spr-price-button/div/div'
        # Find all elements that match the specified XPath pattern
        target_elements = driver.find_elements(By.XPATH, target_xpath_pattern)
        # Extract and print information from all matching elements
        away_card_odds = [target_element.text for target_element in target_elements]

        home_card_df = pd.DataFrame({"card_names": home_card_names, "card_odds": home_card_odds})
        away_card_df = pd.DataFrame({"card_names": away_card_names, "card_odds": away_card_odds})

        cards_df = pd.concat([home_card_df,away_card_df], ignore_index=True)
        cards_df['contest'] = contest_list[i]

        cards_df_list.append(cards_df)
        print('Done with cards')

    time.sleep(3)
    ############################################################
    #Assist click
    ############################################################

    # After clicking the buttons, find buttons with the specified text
    target_text = "Player Assist Minutes"
    # XPath selector for buttons with the target text
    target_buttons_xpath = f'//*[contains(text(), "{target_text}")]'
    # Find all buttons with the target text
    target_buttons3 = driver.find_elements(By.XPATH, target_buttons_xpath)
        # Print or interact with the found buttons
    for j in range(len(target_buttons3)):
        try:
            # Wait up to 10 seconds for the button to be clickable
            WebDriverWait(driver, 10).until(EC.element_to_be_clickable((By.XPATH, target_buttons_xpath)))
            # Scroll the element into view
            ActionChains(driver).move_to_element(target_buttons3[j]).perform()
            # Click the button using JavaScript
            driver.execute_script("arguments[0].click();", target_buttons3[j])
            # Optional: Wait for a moment
            time.sleep(3)
        except TimeoutException:
            print(f"Timeout waiting for the button to be clickable.")
    time.sleep(3)
    ####################################################################
    # Assist x10 click
    ####################################################################
    # After clicking the buttons, find buttons with the specified text
    target_text = "Player Assists x 10"

    # XPath selector for buttons with the target text
    target_buttons_xpath = f'//*[contains(text(), "{target_text}")]'

    # Find all buttons with the target text
    target_buttons4 = driver.find_elements(By.XPATH, target_buttons_xpath)

        # Print or interact with the found buttons
    for j in range(len(target_buttons4)):
        try:
            # Wait up to 10 seconds for the button to be clickable
            WebDriverWait(driver, 10).until(EC.element_to_be_clickable((By.XPATH, target_buttons_xpath)))
            # Scroll the element into view
            ActionChains(driver).move_to_element(target_buttons4[j]).perform()
            # Click the button using JavaScript
            driver.execute_script("arguments[0].click();", target_buttons4[j])
            # Optional: Wait for a moment
            time.sleep(3)
        except TimeoutException:
            print(f"Timeout waiting for the button to be clickable.")
    time.sleep(3)


    ############################################################
    #Assist MARKETS
    ############################################################
    matches = []
    # HOME DATA
    for button in target_buttons3:
        full_xpath = get_full_xpath(button)

        # Regular expression pattern to find "container[<number>]"
        pattern = r"container\[(\d+)\]"

        # Search for all occurrences and extract numbers
        matches = re.findall(pattern, full_xpath)

    if len(matches)!=0:
        # Print extracted numbers
        for match in matches:
            assist_index = match

        init_path = '//sx-pricing-page//sx-p-panel-container['+ str(assist_index)
        #                sx-pricing-page/div[1]/sc-panel/div/div[2]/content/sx-p-panel-container[4]/div/sc-panel/div/div/div[2]/header/div/div[1]/span
        target_xpath_pattern = init_path + ']//sx-p-panel-spreads-column[*]/div/div[*]/div[2]/div/div/sx-p-panel-link-name/div/span'
        retries = 0
        while retries < max_retries:
            try:
                # Find all elements that match the specified XPath pattern
                target_elements = driver.find_elements(By.XPATH, target_xpath_pattern)
                # Extract and print information from all matching elements
                home_AM_names = [target_element.text for target_element in target_elements]

                # Click the element
                1/len(home_AM_names)

                break  # Exit the loop if click is successful

            except ZeroDivisionError:
                print("Zero names Found. Retrying...")
                retries += 1
                # Optional: Wait for a moment
                time.sleep(3)
        home_AM_names

        # After clicking the initial buttons, find elements that match the specified XPath pattern
        target_xpath_pattern = init_path + ']//sx-p-panel-spreads-column[*]/div/div[*]/div[3]/sx-spr-price-button/div/div'
        #                                         sx-p-panel-spreads-column[1]/div/div[2]/div[3]/sx-spr-price-button/div/div
        # Find all elements that match the specified XPath pattern
        target_elements = driver.find_elements(By.XPATH, target_xpath_pattern)
        # Extract and print information from all matching elements
        home_AM_odds = [target_element.text for target_element in target_elements]

        home_assists_df = pd.DataFrame({"AM_names": home_AM_names, "AM_odds": home_AM_odds})
    #     away_assists_df = pd.DataFrame({"AM_names": away_AM_names, "AM_odds": away_AM_odds})

        assists_df = pd.concat([home_assists_df], ignore_index=True) #away_assists_df
        assists_df['contest'] = contest_list[i]
        assists_df_list.append(assists_df)

        print('Done with assists')
    else:
        print('Assists not Found')

    ##################################################################
    # Assists x 10
    ##################################################################
    matches = []
    for button in target_buttons4:
        full_xpath2 = get_full_xpath(button)

        # Regular expression pattern to find "container[<number>]"
        pattern = r"container\[(\d+)\]"

        # Search for all occurrences and extract numbers
        matches = re.findall(pattern, full_xpath2)

    if len(matches)!=0:
        # Print extracted numbers
        for match in matches:
            assist_index2 = match

            init_path = '//sx-pricing-page//sx-p-panel-container['+ str(assist_index2)
    #         target_xpath_pattern = init_path + ']//sx-p-panel-spreads-column[1]/div/div[*]/div[2]/div/div/sx-p-panel-link-name/div/span'
            target_xpath_pattern = init_path + ']//sx-p-panel-spreads-column[*]/div/div[*]/div[2]/div/div/sx-p-panel-link-name/div/span'

            retries = 0
            while retries < max_retries:
                try:
                    # Find all elements that match the specified XPath pattern
                    target_elements = driver.find_elements(By.XPATH, target_xpath_pattern)
                    # Extract and print information from all matching elements
                    home_AM_namesX10 = [target_element.text for target_element in target_elements]

                    # Click the element
                    1/len(home_AM_namesX10)

                    break  # Exit the loop if click is successful

                except ZeroDivisionError:
                    print("Division by zero. Retrying...")
                    retries += 1
                    # Optional: Wait for a moment
                    time.sleep(3)

            # After clicking the initial buttons, find elements that match the specified XPath pattern
            target_xpath_pattern = init_path + ']//sx-p-panel-spreads-column[*]/div/div[*]/div[3]/sx-spr-price-button/div/div'
            # Find all elements that match the specified XPath pattern
            target_elements = driver.find_elements(By.XPATH, target_xpath_pattern)
            # Extract and print information from all matching elements
            home_AM_oddsX10 = [target_element.text for target_element in target_elements]

            home_assistsX10_df = pd.DataFrame({"AM_namesX10": home_AM_namesX10, "AM_oddsX10": home_AM_oddsX10})

            assistsX10_df = pd.concat([home_assistsX10_df], ignore_index=True) #,away_assistsX10_df
            assistsX10_df['contest'] = contest_list[i]
            assistsx10_df_list.append(assistsX10_df)

            print('Done with assistsX10')
    else:
        print('AssistsX10 not Found')

    driver.back()
    time.sleep(3)
```
