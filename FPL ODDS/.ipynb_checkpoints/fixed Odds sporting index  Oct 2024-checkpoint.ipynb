```python
# Function for extracting 
def get_full_xpath(element):
    components = []
    child = element
    while child.tag_name != "html":
        parent = child.find_element(By.XPATH, "..")
        siblings = parent.find_elements(By.XPATH, "*")
        same_tag_siblings = [sib for sib in siblings if sib.tag_name == child.tag_name]

        # Attempt to use a more stable identifier if available
        data_id = child.get_attribute("data-id")
        if data_id:
            position = f"[@data-id='{data_id}']"
        elif len(same_tag_siblings) > 1:
            index = same_tag_siblings.index(child) + 1
            position = f"[{index}]"
        else:
            position = ''

        components.append(f"{child.tag_name}{position}")
        child = parent

    components.append("html")
    components.reverse()
    return '/' + '/'.join(components)
```

```python
# Function to determine if a scoreline is a draw
def is_draw(scoreline):
    home_goals, away_goals = map(int, scoreline.split('-'))
    return home_goals == away_goals
```

```python
# Define the odds with the corresponding scores for each team and draws
west_ham_odds_extended_with_scores = [
    (1, 0, 16), (2, 0, 35), (2, 1, 18), (3, 0, 90), (3, 1, 55), (3, 2, 55), (4, 0, 200), 
    (4, 1, 125), (4, 2, 125), (4, 3, 150), (5, 0, 250), (5, 1, 250), (5, 2, 200), 
    (6, 0, 250), (6, 1, 250), (7, 0, 250), (8, 0, 250)
]

arsenal_odds_extended_with_scores = [
    (0, 1, 7), (0, 2, 13/2), (1, 2, 15/2), (0, 3, 11), (1, 3, 12), (2, 3, 25), 
    (0, 4, 22), (1, 4, 45), (2, 4, 100), (0, 5, 50), (1, 5, 50), (2, 5, 90), 
    (0, 6, 100), (1, 6, 100), (0, 7, 200), (0, 8, 250)
]

draw_odds_extended_with_scores = [
    (0, 0, 14), (1, 1, 15/2), (2, 2, 16), (3, 3, 70), (4, 4, 200)
]

# Convert odds to probabilities while accounting for the scores
def convert_odds_with_scores(odds_with_scores):
    probabilities_with_scores = []
    for home_score, away_score, odds in odds_with_scores:
        probability = 1 / (odds + 1)  # Convert odds to probability
        probabilities_with_scores.append((home_score, away_score, probability))
    return probabilities_with_scores

# Apply the conversion to all sets of odds
west_ham_probabilities_with_scores = convert_odds_with_scores(west_ham_odds_extended_with_scores)
arsenal_probabilities_with_scores = convert_odds_with_scores(arsenal_odds_extended_with_scores)
draw_probabilities_with_scores = convert_odds_with_scores(draw_odds_extended_with_scores)

# Calculate the total implied probability for normalization
total_implied_probability = sum(prob for _, _, prob in west_ham_probabilities_with_scores) + \
                            sum(prob for _, _, prob in arsenal_probabilities_with_scores) + \
                            sum(prob for _, _, prob in draw_probabilities_with_scores)

# Normalize the probabilities
west_ham_normalized_probs = [(home_score, away_score, prob / total_implied_probability)
                             for home_score, away_score, prob in west_ham_probabilities_with_scores]
arsenal_normalized_probs = [(home_score, away_score, prob / total_implied_probability)
                            for home_score, away_score, prob in arsenal_probabilities_with_scores]
draw_normalized_probs = [(home_score, away_score, prob / total_implied_probability)
                         for home_score, away_score, prob in draw_probabilities_with_scores]

# Aggregate the probabilities for each number of goals scored by each team
west_ham_goals_probs = {}
arsenal_goals_probs = {}

# Aggregate probabilities from all possible outcomes
for home_score, away_score, prob in west_ham_normalized_probs + draw_normalized_probs:
    west_ham_goals_probs[home_score] = west_ham_goals_probs.get(home_score, 0) + prob
    arsenal_goals_probs[away_score] = arsenal_goals_probs.get(away_score, 0) + prob

for home_score, away_score, prob in arsenal_normalized_probs:
    west_ham_goals_probs[home_score] = west_ham_goals_probs.get(home_score, 0) + prob
    arsenal_goals_probs[away_score] = arsenal_goals_probs.get(away_score, 0) + prob

# Calculate the expected goals for each team
expected_goals_west_ham = sum(score * prob for score, prob in west_ham_goals_probs.items())
expected_goals_arsenal = sum(score * prob for score, prob in arsenal_goals_probs.items())

expected_goals_west_ham, expected_goals_arsenal
```
