```python
from selenium import webdriver
from selenium.webdriver.common.by import By
import pandas as pd
import time
```

```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from webdriver_manager.chrome import ChromeDriverManager
from selenium.webdriver.common.by import By
```

```python
driver_path = r"C:\Users\burio\Downloads\chromedriver-win64\chromedriver.exe"
s = Service(driver_path)
driver = webdriver.Chrome(service=s)
```

```python
url = "https://members.fantasyfootballscout.co.uk/h2h/league/2024/2/"
```

```python
# Open the webpage
driver.get(url)

# Wait for the page to load completely
time.sleep(5)  # Adjust sleep time as necessary

# Handle the cookie consent bar if present
try:
    cookie_accept_button = driver.find_element(By.ID, "cookie-bar")  # Replace with the actual element selector
    cookie_accept_button.click()
    time.sleep(2)  # Wait for the cookie consent bar to disappear
except Exception as e:
    print("No cookie consent button found or clicked.")

# Initialize a list to store all the data
all_data = []

# Function to extract data from the current page
def extract_data():
    table_body = driver.find_element(By.TAG_NAME, 'tbody')
    for row in table_body.find_elements(By.TAG_NAME, 'tr'):
        cells = row.find_elements(By.TAG_NAME, 'td')
        cells_text = [cell.text for cell in cells]
        all_data.append(cells_text)

# Extract data from the first page
extract_data()

# Loop to navigate through all the pages
while True:
    try:
        # Find the 'next' button using its full XPath
        next_button = driver.find_element(By.XPATH, '/html/body/div/div/div[3]/div/div[1]/div[2]/div/div/div[2]/div[2]/a[3]')
        # Check if the 'next' button is enabled (not disabled)
        if 'disabled' not in next_button.get_attribute('class'):
            next_button.click()
            time.sleep(3)  # Wait for the page to load
            extract_data()
        else:
            break
    except Exception as e:
        print("Exception:", e)
        break

# Close the WebDriver
driver.quit()

# Convert the data into a DataFrame
df = pd.DataFrame(all_data, columns=["Name", "League", "Division", "Team id", "points", "OR"])

# Display the DataFrame
print(df)

# Save the DataFrame to a CSV file
df.to_csv('player_info_all_pages.csv', index=False)
```
